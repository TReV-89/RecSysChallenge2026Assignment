```python
from pathlib import Path
import pandas as pd
import pyarrow.parquet as pq

data_dir = Path("data")
if not data_dir.is_dir():
    data_dir = Path("prepared-dataset-eda/data")
if not data_dir.is_dir():
    raise FileNotFoundError(f"Could not find prepared data directory from {Path.cwd()}")

for path in sorted(data_dir.glob("*.parquet")):
    schema = pq.read_schema(path)
    print(f"\n{path.name}")
    print("Columns:", schema.names)

    df = pd.read_parquet(path)
    print("Rows and columns:", df.shape)
    display(df.head(3))
```

```python
sessions = pd.read_parquet(data_dir / "sessions.parquet")
interactions = pd.read_parquet(data_dir / "conversation_interactions.parquet")
queries = pd.read_parquet(data_dir / "turn_queries.parquet")

display(sessions.dtypes)
display(sessions.isna().sum().sort_values(ascending=False))

display(interactions.head())
display(queries.head())
```
